## Step 1: Import libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## Step 2: Load dataset

```python
from google.colab import files
uploaded = files.upload()
dataset = pd.read_csv("emails.csv")
```

## Step 4: Remove missing values

```python
dataset = dataset.dropna()
dataset.shape
```

## Step 5: Separate features and target

```python
X = dataset.drop(columns=["Email No.", "Prediction"], errors="ignore")
y = dataset["Prediction"]
```

## Step 6: Split into training and testing data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1, stratify=y
)
```

## Step 7: Scale features

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Step 8: Train KNN model

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
```

## Step 9: Predict using KNN

```python
y_pred_knn = knn.predict(X_test)
```

## Step 10: Evaluate KNN

```python
accuracy_score(y_test, y_pred_knn)
```

```python
confusion_matrix(y_test, y_pred_knn)
```

```python
classification_report(y_test, y_pred_knn)
```

## Step 11: Train SVM model

```python
svm = SVC(kernel="linear")
svm.fit(X_train, y_train)
```

## Step 12: Predict using SVM

```python
y_pred_svm = svm.predict(X_test)
```

## Step 13: Evaluate SVM

```python
accuracy_score(y_test, y_pred_svm)
```

```python
confusion_matrix(y_test, y_pred_svm)
```

```python
classification_report(y_test, y_pred_svm)
```

## Step 14: Compare both models

```python
knn_accuracy = accuracy_score(y_test, y_pred_knn)
svm_accuracy = accuracy_score(y_test, y_pred_svm)

comparison = pd.DataFrame({
    "Model": ["KNN", "SVM"],
    "Accuracy": [knn_accuracy, svm_accuracy]
})

comparison
```

## Step 15: Best model based on accuracy

```python
if knn_accuracy > svm_accuracy:
    display("KNN performs better based on accuracy.")
elif svm_accuracy > knn_accuracy:
    display("SVM performs better based on accuracy.")
else:
    display("Both models have equal accuracy.")
```
